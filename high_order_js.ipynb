```python
%%writefile code.js
// Part A

const data = [5, 12, 3, 20, 11, 9, 30];

const l = [];
for (let i = 0; i < data.length; i++) {
  if (data[i] > 10) {
    l.push(data[i] * 2);
  }
}

const pipelineResult = data.filter(x => x > 10).map(x => x * 2);

console.log("Part A");
console.log("Loop result:", l);
console.log("Pipeline result:", pipelineResult);
console.log("Results match:", JSON.stringify(l) === JSON.stringify(pipelineResult));
console.log();

// Part B

const numbers = [1, 2, 3, 4, 5, 6];

const s = numbers.filter(n => n % 2 === 0).map(n => n * 2).reduce((sum, n) => sum + n, 0);

console.log("Part B");
console.log("Sum of doubled evens:", s);
console.log();

// Part C

const users = [
  { id: 1, name: "Alice", age: 25 },
  { id: 2, name: "Bob", age: 17 },
  { id: 3, name: "Carol", age: 19 }
];

console.log("Part C");

const a = users.some(user => user.age < 18);
console.log("Any user under 18?", a);

const w = users.find(user => user.id === 3);
console.log("User with id 3:", w);

console.log("User names:");
users.forEach(user => console.log(user.name));
console.log();

// Part D

function repeat(n, action) {
  for (let i = 0; i < n; i++) {
    action(i);
  }
}

console.log("Part D");
console.log("repeat(3, console.log):");
repeat(3, console.log);
console.log("repeat(4, i => console.log('Step ' + i)):");
repeat(4, i => console.log("Step " + i));
```
